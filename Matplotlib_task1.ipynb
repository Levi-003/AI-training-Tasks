```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("tvshows.csv")
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
df["year"] = df["date_added"].dt.year
data = df.groupby(["year", "type"]).size().unstack()
data = data.loc[2015:2021]
print(data)
```

```python
ax = data.plot(
    kind="bar",
    color=["red", "blue"],
    figsize=(8, 5)
)
for bars in ax.containers:
    ax.bar_label(bars)
total = data["Movie"] + data["TV Show"]
average = total.mean()
ax.axhline(
    average,
    color="black",
    linestyle="dotted",
    label="Average"
)
plt.title("Netflix movies added per year")
plt.xlabel("Year")
plt.ylabel("No of movies")
plt.grid(axis="y", linestyle="dotted")
plt.legend()
plt.show()
```

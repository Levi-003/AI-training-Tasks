```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

df = pd.read_csv(list(uploaded.keys())[0], encoding="latin1")

df.head(10)
```

```python
print(df.columns.tolist())
```

```python
print("1. First 10 Rows:")
print(df.head(10))

print("\n2. Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\n3. Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\n4. Summary Statistics:")
print(df.describe())

print("\n5. Unique Countries:")
print(df["Country name"].nunique())

print("\n6. Country with Highest Happiness Score:")
happiest = df.loc[df["Life Ladder"].idxmax()]
print("Country:", happiest["Country name"])
print("Happiness Score:", happiest["Life Ladder"])

print("\n7. Most Common Region:")
print("Region information is not available in this dataset.")
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Life Ladder"], bins=10, edgecolor="black")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")
plt.title("Distribution of Happiness Scores")
plt.show()

top10 = df.groupby("Country name")["Life Ladder"].mean().sort_values(ascending=False).head(10)

print("Top 10 Happiest Countries:")
print(top10)

plt.figure(figsize=(10,5))
plt.bar(top10.index, top10.values)
plt.xticks(rotation=45, ha="right")
plt.xlabel("Country")
plt.ylabel("Average Happiness Score")
plt.title("Top 10 Happiest Countries")
plt.show()

bottom10 = df.groupby("Country name")["Life Ladder"].mean().sort_values().head(10)

print("\nBottom 10 Countries:")
print(bottom10)

plt.figure(figsize=(10,5))
plt.bar(bottom10.index, bottom10.values)
plt.xticks(rotation=45, ha="right")
plt.xlabel("Country")
plt.ylabel("Average Happiness Score")
plt.title("Bottom 10 Countries")
plt.show()

plt.figure(figsize=(8,5))
plt.scatter(df["Log GDP per capita"], df["Life Ladder"])
plt.xlabel("Log GDP per Capita")
plt.ylabel("Happiness Score")
plt.title("GDP per Capita vs Happiness Score")
plt.show()

plt.figure(figsize=(8,5))
sns.boxplot(y=df["Life Ladder"])
plt.ylabel("Happiness Score")
plt.title("Distribution of Happiness Scores")
plt.show()
```

```python
mean_score = df["Life Ladder"].mean()
median_score = df["Life Ladder"].median()
mode_score = df["Life Ladder"].mode()[0]

print("Mean:", mean_score)
print("Median:", median_score)
print("Mode:", mode_score)

variance = df["Life Ladder"].var()
std = df["Life Ladder"].std()

print("\nVariance:", variance)
print("Standard Deviation:", std)

global_average = df["Life Ladder"].mean()
below_average = df[df["Life Ladder"] < global_average]

percentage = (len(below_average) / len(df)) * 100

print("\nPercentage below global average:", percentage, "%")

top20 = df.sort_values("Life Ladder", ascending=False).head(20)

print("\nMost Common Region among Top 20:")
print("Region information is not available in this dataset.")
```

```python
happiness = df["Life Ladder"].to_numpy()
gdp = df["Log GDP per capita"].to_numpy()

print("Happiness Array:")
print(happiness)

print("\nGDP Array:")
print(gdp)

addition = happiness + gdp
subtraction = happiness - gdp

print("\nVector Addition:")
print(addition)

print("\nVector Subtraction:")
print(subtraction)

dot_product = np.dot(happiness, gdp)

print("\nDot Product:")
print(dot_product)

matrix = df[[
    "Log GDP per capita",
    "Social support"
]].to_numpy()

weights = np.array([0.5, 0.5])

result = np.dot(matrix, weights)

print("\nMatrix Multiplication:")
print(result)

normalized = (
    happiness - happiness.min()
) / (
    happiness.max() - happiness.min()
)

print("\nNormalized Happiness Score:")
print(normalized)
```

```python
df["Wellbeing_Index"] = (
    df["Log GDP per capita"] * df["Life Ladder"]
) + 0.5 * (df["Life Ladder"] - 5) ** 2

df["Derivative"] = (
    df["Log GDP per capita"] + df["Life Ladder"] - 5
)

print(df[[
    "Country name",
    "Life Ladder",
    "Log GDP per capita",
    "Wellbeing_Index",
    "Derivative"
]].head())
```

```python
df["is_happy"] = np.where(df["Life Ladder"] > 7, 1, 0)

df["happiness_rank"] = df["Life Ladder"].rank(
    ascending=False,
    method="min"
)

gdp_mean = df["Log GDP per capita"].mean()

df["above_average_gdp"] = np.where(
    df["Log GDP per capita"] > gdp_mean,
    1,
    0
)

df["life_expectancy_quartile"] = pd.qcut(
    df["Healthy life expectancy at birth"],
    4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

print(df[[
    "Country name",
    "Life Ladder",
    "is_happy",
    "happiness_rank",
    "Log GDP per capita",
    "above_average_gdp",
    "Healthy life expectancy at birth",
    "life_expectancy_quartile"
]].head(10))
```

```python
happiest = df.loc[df["Life Ladder"].idxmax()]

print("1. Happiest Country:")
print(happiest["Country name"])
print("Happiness Score:", happiest["Life Ladder"])

happiest_year = df.groupby("year")["Life Ladder"].mean().idxmax()

print("\n2. Year with Highest Average Happiness:")
print(happiest_year)

correlation = df["Log GDP per capita"].corr(
    df["Life Ladder"]
)

print("\n3. GDP-Happiness Correlation:")
print(correlation)

if correlation > 0.7:
    print("GDP and happiness have a strong positive correlation.")
elif correlation > 0.4:
    print("GDP and happiness have a moderate positive correlation.")
else:
    print("GDP and happiness have a weak positive correlation.")

print("\n4. Do richer countries always have higher happiness scores?")
print("No. Richer countries generally tend to have higher happiness, but GDP does not always determine happiness.")

life_year = df.groupby("year")["Healthy life expectancy at birth"].mean().idxmax()

print("\n5. Year with Highest Average Life Expectancy:")
print(life_year)
```

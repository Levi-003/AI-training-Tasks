```python
arr = np.array([10, np.nan, 20, 30, np.nan, 40])

missing = np.isnan(arr)

print(missing)
```

```python
count = np.isnan(arr).sum()
print("Missing values:", count)
```

```python
result = np.nan_to_num(arr, nan=0)
print(result)
```

```python
mean = np.nanmean(arr)
result = np.where(np.isnan(arr), mean, arr)
print(result)
```

```python
arr2 = np.array([
    [1, 2],
    [3, np.nan],
    [5, 6],
    [np.nan, np.nan]
], dtype=float)

result = arr2[~np.isnan(arr2).any(axis=1)]

print(result)
```

```python
arr = np.array([10, np.nan, np.nan, 40, 50])

x = np.arange(len(arr))
valid = ~np.isnan(arr)

result = np.interp(x, x[valid], arr[valid])

print(result)
```

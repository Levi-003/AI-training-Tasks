```python
df = pd.read_csv(
    "communities.data",
    header=None,
    na_values="?"
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
with open("communities.names", "r") as file:
    names_text = file.read()

print(names_text[:5000])
```

```python
import re

attributes = re.findall(
    r'@attribute\s+(\S+)\s+',
    names_text,
    flags=re.IGNORECASE
)

print("Number of column names:", len(attributes))

print("\nColumn names:")
for i, col in enumerate(attributes):
    print(i, col)
```

```python
df.columns = attributes

print("Column names assigned successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0].sort_values(
    ascending=False
)

print("Columns with missing values:")
display(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Calculate missing percentage
missing_ratio = df.isnull().mean()

# Select columns with more than 50% missing values
high_missing_columns = missing_ratio[
    missing_ratio > 0.50
].index

print("Columns removed:")

for col in high_missing_columns:
    print(col)

# Remove them
df_clean = df.drop(
    columns=high_missing_columns
)

print("\nOriginal shape:", df.shape)
print("Shape after removing high-missing columns:", df_clean.shape)
```

```python
columns_to_remove = [
    "state",
    "county",
    "community",
    "communityname",
    "fold",
    "ViolentCrimesPerPop"
]

df_clean = df_clean.drop(
    columns=columns_to_remove,
    errors="ignore"
)

print("Shape after removing non-feature columns and target:")
print(df_clean.shape)
```

```python
# Fill remaining numerical missing values with median
df_clean = df_clean.fillna(
    df_clean.median(numeric_only=True)
)

print("Remaining missing values:")
print(df_clean.isnull().sum().sum())
```

```python
print("Cleaned dataset shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum().sum())

print("\nData types:")
print(df_clean.dtypes)

display(df_clean.head())
```

```python
plt.figure(figsize=(15, 8))

df_clean.hist(
    figsize=(15, 10),
    bins=20
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 12))

correlation_matrix = df_clean.corr()

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(df_clean)

print("Original dimensions:", df_clean.shape)
print("Standardized dimensions:", X_scaled.shape)

print("\nFirst 5 standardized rows:")
display(pd.DataFrame(
    X_scaled,
    columns=df_clean.columns
).head())
```

```python
pca = PCA()

X_pca = pca.fit_transform(X_scaled)

print("Original number of features:", X_scaled.shape[1])
print("Number of PCA components:", X_pca.shape[1])
```

```python
pca = PCA()

X_pca = pca.fit_transform(X_scaled)

print("Original number of features:", X_scaled.shape[1])
print("Number of PCA components:", X_pca.shape[1])
```

```python
explained_variance = pca.explained_variance_ratio_

print("Explained Variance Ratio:\n")

for i, variance in enumerate(explained_variance):
    print(
        f"PC{i+1}: {variance:.4f} "
        f"({variance * 100:.2f}%)"
    )
```

```python
cumulative_variance = np.cumsum(
    explained_variance
)

print("Cumulative Explained Variance:\n")

for i, variance in enumerate(cumulative_variance):
    print(
        f"PC{i+1}: {variance:.4f} "
        f"({variance * 100:.2f}%)"
    )
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    range(1, len(explained_variance) + 1),
    explained_variance,
    marker="o"
)

plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.title("Explained Variance by Principal Component")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o"
)

plt.axhline(
    y=0.95,
    linestyle="--",
    label="95% Variance"
)

plt.axvline(
    x=35,
    linestyle="--",
    label="35 Components"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance")

plt.legend()
plt.grid(True)
plt.show()
```

```python
n_components = 35

pca_final = PCA(
    n_components=n_components
)

X_pca_reduced = pca_final.fit_transform(
    X_scaled
)

print("Original dimensions:", X_scaled.shape)
print("Reduced dimensions:", X_pca_reduced.shape)
```

```python
pca_columns = [
    f"PC{i+1}"
    for i in range(n_components)
]

pca_df = pd.DataFrame(
    X_pca_reduced,
    columns=pca_columns
)

print("PCA transformed dataset:")
display(pca_df.head())

print("\nPCA dataset shape:", pca_df.shape)
```

```python
selected_variance = pca_final.explained_variance_ratio_

print("Explained variance by selected components:\n")

for i, variance in enumerate(selected_variance):
    print(
        f"PC{i+1}: {variance * 100:.2f}%"
    )

print(
    "\nTotal variance retained:",
    f"{selected_variance.sum() * 100:.2f}%"
)
```

```python
plt.figure(figsize=(10, 7))

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    alpha=0.6
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA - Reduced Dimensional Data")

plt.grid(True)
plt.show()
```

```python
original_dimensions = df_clean.shape[1]
reduced_dimensions = pca_df.shape[1]

comparison = pd.DataFrame({
    "Dataset": [
        "Original",
        "After PCA"
    ],
    "Number of Features": [
        original_dimensions,
        reduced_dimensions
    ]
})

display(comparison)

reduction_percentage = (
    (original_dimensions - reduced_dimensions)
    / original_dimensions
) * 100

print(
    f"Dimensionality reduced by: "
    f"{reduction_percentage:.2f}%"
)
```

```python
loadings = pd.DataFrame(
    pca_final.components_.T,
    columns=pca_columns,
    index=df_clean.columns
)

print("PCA Component Loadings:")
display(loadings)
```

```python
for component in pca_columns[:5]:

    print(f"\nTop features contributing to {component}:")

    top_features = (
        loadings[component]
        .abs()
        .sort_values(ascending=False)
        .head(5)
    )

    print(top_features)
```

```python
print("========== PCA FINAL SUMMARY ==========")

print(f"Original number of features: {original_dimensions}")
print(f"Reduced number of features: {reduced_dimensions}")

print(
    f"Variance retained: "
    f"{selected_variance.sum() * 100:.2f}%"
)

print(
    f"Dimensionality reduction: "
    f"{reduction_percentage:.2f}%"
)

print("\nInterpretation:")
print(
    f"The original {original_dimensions} numerical features "
    f"were reduced to {reduced_dimensions} principal components."
)

print(
    f"These {reduced_dimensions} components retain approximately "
    f"{selected_variance.sum() * 100:.2f}% of the total variance."
)

print(
    "Therefore, PCA significantly reduces the dimensionality "
    "while preserving most of the information in the dataset."
)
```
